```python
import re
```

```python
def valid_username(username):
    pattern = r"\w{3,15}"
    return re.fullmatch(pattern, username) is not None

print(valid_username("john_123"))
print(valid_username("ab"))
print(valid_username("john smith"))
```

```python

pattern = r"cat|dog"

print(re.fullmatch(pattern, "cat"))
print(re.fullmatch(pattern, "dog"))
print(re.fullmatch(pattern, "cow"))
print(re.fullmatch(pattern, "cat123"))
```
