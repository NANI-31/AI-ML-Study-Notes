```python
import re
```

```python
text = "I have 123 apples and b1234b oranges."

result = re.findall(r"\w+\d+\w+", text)

print(result)
```

```python
print(re.findall(r"^abc\w*", "abccdef"))
```

```python
print(re.findall(r"[^abc]+", "abcdef"))
```

```python
text = "https://example.com is my website"

result = re.search(r"^https?://", text)

print(result)
```
