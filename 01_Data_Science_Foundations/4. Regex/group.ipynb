```python
import re
```

```python
def p(r):
    print(r)
    if r is None:
        return

    if isinstance(r, (list, str)):
        # print(f"findall items ({len(r)}) -> {r}")
        return

    if r.groupdict():
        print("groups ->", r.groups())
        print("named groups ->", r.groupdict())
        for name, val in r.groupdict().items():
            print(f"group:{name} -> {val}")
        return

    # for i in range(1, len(r.groups()) + 1):
    #     print(f"group:{i} -> {r.group(i)}")
    if r.groups():
        print("groups ->", r.groups())
        for i, val in enumerate(r.groups(), start=1):
            print(f"group:{i} -> {val}")

```

```python
match = re.search(r"(a)(b)(c)", "abc")
p(match)
```

```python
values = ["123-456", "-456"]
for val in values:
    r = re.fullmatch(r"(\d{3})?-(\d{3})", val)
    p(r)
```

```python
values = ["123-456", "456"]
for val in values:
    r = re.fullmatch(r"(?P<area>\d{3})?-?(?P<code>\d{3})", val)
    p(r)
```

```python
text = "123, 344, 54"
t1 = "cat cat123 dog"
r = re.findall(r"\d+", text)
p(r)
r = re.findall(r"(cat)(\d+)", t1)
p(r)
```

```python
r = re.search(r"(a)(b)(c)", "abc")
p(r)
```

```python
r = re.search(r"((a)(b)(c))", "abc")
p(r)
```

```python
r = re.search(r"(?:a)(b)(c)", "abc")
p(r)
```

```python
r = re.search(r"(?:(?:a)(b)(c))", "abc")
p(r)
```

```python
result = re.sub(
    r"(\w+)\s+(\w+)",
    r"\2 \1",
    text
)
p(result)
```

```python
text = "hello hello world world"
pattern = r"\b(\w+)\s+\1\b"

result = re.findall(pattern, text)

print(result)
```

```python
text = "<div class=`hello` id=`1`>Hello <p> world </p> </div>"

r = re.findall(r"<([A-Za-z][A-Za-z0-9]*)[^>]*>", text)
p(r)
```

```python
t = "chun@gmail.com"
r = re.findall(r"[^@]+@.*", t)
p(r)
```
