````python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 25, 30],
    "Marks": [85, 90, 95]
})

df
````

# Why convert DataFrame to NumPy?
```
Pandas and NumPy are designed for slightly different purposes.

Pandas
Good for working with structured/tabular data:
df["Age"]
df["Marks"]
df[df["Marks"] > 90]

NumPy
Good for numerical computation and arrays:
array.mean()
array.shape
array[0, 1]

For example, if you have a machine-learning algorithm that expects a NumPy array, you might convert:

X = df.to_numpy()

```
