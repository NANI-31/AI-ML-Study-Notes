```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita", "Kiran"],
    "Age": [24, 25, 23, 26, 24],
    "Salary": [50000, 65000, 55000, 70000, 60000],
    "Branch": ["CSE", "ECE", "CSE", "ECE", "CSE"]
}, index=["A", "B", "C", "D", "E"])
df
```

```python
df.iloc[[-3,-2],[1, 2] ]
```

### **Specific Rows and Columns**

`df.loc[rows, columns]`

`df.iloc[rows, columns]`

```python
df.loc[["B", "D"], ["Name", "Salary"]]
df.iloc[[0, 2], [0, 2]]
```

### **Selecting Ranges**

```python
df.iloc[0:3]
df.iloc[0:3, 0:2]
```
