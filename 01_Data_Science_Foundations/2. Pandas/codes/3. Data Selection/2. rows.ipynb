### **Selecting Rows by Index**

`ending label included`

```python
print(df.loc[0]) # single row
print(df.loc[[0,2,4]]) # multi custom row
print(df.loc[0:2]) # multi sequence row 
```

### **1. Using `loc[]`**
`loc[]` performs label-based selection.

current index is labeled as `["A","B",...]`

```python
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita", "Kiran"],
    "Age": [24, 25, 23, 26, 24],
    "Salary": [50000, 65000, 55000, 70000, 60000]
}, index=["A", "B", "C", "D", "E"])
df
```

```python
print(df.loc["C"])
df.loc[["A", "C", "E"]]
print(df.loc["A":"C"])
```

`df.loc[0:2]`

pandas looks for index labels `0` through `2`. But your index contains strings, not integers, so you get:

`TypeError: cannot do slice indexing on Index with these indexers [0] of type int`

### **3. Using `iloc[]`**
`iloc[]` means integer-location based indexing.

It uses the position of the row.

|Position|Label|
|--|--|
   |0|           A|
   1           |B|
   2           |C|
   3           |D|
   4           |E|

with `iloc`, the ending position is excluded

```python
print(df.iloc[0])
print(df.iloc[0:3])
print(df.iloc[[0, 2, 4]])
```

`iloc` → ending position excluded

`loc`  → ending label included

### **4. `tail()`**
