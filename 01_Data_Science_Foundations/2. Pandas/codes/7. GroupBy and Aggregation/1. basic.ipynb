```python
import pandas as pd
df = pd.DataFrame({
    "Name": [
        "Ravi", "Priya", "Arun",
        "Sita", "Kiran", "Meena"
    ],
    "Department": [
        "IT", "HR", "IT",
        "HR", "IT", "Sales"
    ],
    "Salary": [
        50000, 60000, 55000,
        70000, 65000, 80000
    ],
    "Experience": [
        2, 4, 3, 6, 5, 7
    ]
})
df
```

```python
result = df.groupby("Department")["Salary"].mean().reset_index()
result
```

```python
# df
df.groupby("Department")["Salary"].agg(
    ["mean", "max", "min"]
).reset_index()
```

```python
df = pd.DataFrame({
    "Region": [
        "South", "South", "North",
        "North", "South", "North"
    ],
    "Product": [
        "Laptop", "Phone", "Laptop",
        "Phone", "Laptop", "Laptop"
    ],
    "Sales": [
        100000, 50000, 120000,
        60000, 90000, 110000
    ]
})
# 	What are the total sales for each product in each region?
```

```python
df.groupby(["Region", "Product"])["Sales"].sum().reset_index()
```
