```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita"],
    "Age": [24, 25, 23, 26],
    "Salary": [50000, 65000, 55000, 70000],
    "Branch": ["CSE", "ECE", "CSE", "ECE"]
})
print(df)
```

# Columns

```python
df["Experience"] = [1, 3, 2, 4]
df
```

```python
df["Country"] = "India"
df
```

```python
df["Annual_Salary"] = df["Salary"] * 12
df
```

```python
df["Age_Group"] = df["Age"].apply(
    lambda age: "Young" if age < 25 else "Adult"
)
df
```

```python
df["Salary_per_Age"] = (
    df["Salary"] / df["Age"]
)
df
```

## assign()

```python
df = df.assign(
    Annual_Salary=df["Salary"] * 12,
    Salary_Lakhs=df["Salary"] / 100000
)
df
```

```python
df = (
    df.assign(
        Annual_Salary=lambda x: x["Salary"] * 12,
        Salary_Lakhs=lambda x: x["Salary"] / 100000
    )
)
df
```

```python
new_row = pd.DataFrame({
    "Name": ["Kiran"],
    "Age": [24],
    "Salary": [60000],
    "Branch": ["CSE"]
})
```

```python
df = pd.concat([df, new_row], ignore_index=True)
df
```

```python
new_rows = pd.DataFrame({
    "Name": ["Kiran", "Meena"],
    "Age": [24, 27],
    "Salary": [60000, 75000],
    "Branch": ["CSE", "ECE"]
})
df = pd.concat(
    [df, new_rows],
    ignore_index=True
)
df
```

```python
df = pd.DataFrame({
    "Name": ["John", "Alice", "Bob", "Sara"],
    "Age": [20, 30, 22, 40]
})
```

```python
df["Where_Result"] = df["Age"].where(df["Age"] < 25, "Adult")
df
```

```python
df["Mask_Result"] = df["Age"].mask(df["Age"] < 25, "Young")
df
```

`where()` → keep when True

`mask()` → replace when True

`np.where()` → choose between two values
