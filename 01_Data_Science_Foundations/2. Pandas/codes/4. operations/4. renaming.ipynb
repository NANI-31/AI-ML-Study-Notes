```python
import pandas as pd

df = pd.DataFrame({
    "Employee Name ": ["Ravi", "Priya", "Arun", "Ravi", "Sneha"],
    "Employee Age": [24, 25, 23, 24, 28],
    "Monthly Salary": [50000, 65000, 55000, 50000, 75000],
    "Department Name": ["Computer Science", "Electronics", "Computer Science", "Computer Science", "Mechanical"],
    "Annual Salary ($)": [600000, 780000, 660000, 600000, 900000]
})
df
```

### **1. Renaming Columns**

```python
df.columns = [
    "name",
    "age",
    "salary",
    "branch"
]
```

### **2. Renaming Indexes**

```python
df.index = ["S1", "S2", "S3", "S4"]
```

### **3. Using `rename()`**

```python
df = df.rename(
    columns={"Salary": "Annual_Salary"}
)

df = df.rename(
    columns={
        "Name": "Employee_Name",
        "Age": "Employee_Age",
        "Salary": "Monthly_Salary"
    }
)
```

```python
df = df.rename(
    index={
        0: "A",
        1: "B",
        2: "C"
    }
)
```

### **Renaming with a function**

```python
df = df.rename(
    columns=str.lower
)
```

### **4. Standardizing Column Names**
You might receive
"First Name"
A common standard is:
`snake_case`
first_name


```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

### **Handling special characters**
"Annual Salary ($)"

annual_salary

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-z0-9]+", "_", regex=True)
    .str.strip("_")
)
```
