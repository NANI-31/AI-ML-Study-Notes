```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita"],
    "Age": [24, 25, 23, 26],
    "Salary": [50000, 65000, 55000, 70000],
    "Branch": ["CSE", "ECE", "CSE", "ECE"]
})
print(df)
```

### **Single Cell**

```python
df.loc[0, "Salary"] = 55000
df.loc[1, "Branch"] = "CSE"
df.iloc[0, 2] = 55000
```

## **Entire Col**

```python
df["Salary"] = df["Salary"] * 1.10
```

### Replacing a column with new values

```python
df["Branch"] = ["CSE", "CSE", "CSE", "ECE"]
```

### **Updating Values Using Conditions**

```python
df.loc[df["Salary"] < 60000, "Salary"] *= 1.10
df.loc[df["Age"] < 0, "Age"] = None #ages should not be negative.
```

### **Using `replace()`**

```python
print(type(df["Branch"]))
```

```python
df["Branch"] = df["Branch"].replace(
    "Computer Science",
    "CSE"
)
df["Branch"] = df["Branch"].replace({
    "Computer Science": "CSE",
    "Electronics": "ECE"
})
print(df["Branch"].unique())
df["Branch"]
```

#  **Updating Rows**

```python
df.loc[0] = ["Rahul", 25, 60000, "CSE"]
```

### **Updating selected columns in a row**

```python
df.loc[0, ["Age", "Salary"]] = [26, 65000]
```
