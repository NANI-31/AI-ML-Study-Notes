### **3. method**
It decides what happens when two or more values are tied.

`method="average"` — default

````python
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sneha"],
    "Salary": [50000, 70000, 50000, 60000]
})
````

````python
# avg rank  
df["Salary"].rank()
````

````python
df["Salary"].rank(method="min")
df["Salary"].rank(method="max")
# This gives ranks based on the order in which the values appear.
df["Salary"].rank(method="first")
# Tied values get the same rank, but the next rank doesn't skip.
df["Salary"].rank(method='dense')
````
