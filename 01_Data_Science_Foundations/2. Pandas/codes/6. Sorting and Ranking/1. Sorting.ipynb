````python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita", "Kiran", "Meena", "Siva"],
    "Branch": ["CSE", "ECE", "CSE", "ECE", "CSE", "IT", "CSE"],
    "Age": [24, 25, 23, 26, 24, 29, 28],
    "Marks": [78, 92, 85, 92, 67, 88, 92],
    "Salary": [50000, 65000, 55000, 70000, 60000, 80000, 900000]
})
````

### **1. sort_values()**
Sorting doesn't normally change the original DataFrame

sorts based on column values

```ruby
df.sort_values(
    by="column_name"
    axis=0,
    ascending=True,
    inplace=False,
    kind="quicksort",
    na_position="last",
    ignore_index=False,
    key=None
)


````python
df.sort_values(by="Marks", ascending=False, inplace=False)
````

````python
df.sort_values(by=['Marks', 'Branch'], ascending=[False, True])
````

## **key**

key allows you to transform the values before sorting.

````python
df = pd.DataFrame({
    "Name": ["ravi", "PRIYA", "arun", "SNEHA"]
})
df
````

````python
df.sort_values(
    by="Name",
    key=lambda x: x.str.lower()
)
````

### **Sorting by Index**
sorts based on the index labels

```ruby
df.sort_index(
    axis=0,
    level=None,
    ascending=True,
    inplace=False,
    kind="quicksort",
    na_position="last",
    sort_remaining=True,
    ignore_index=False,
    key=None
)
```

````python
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sneha"],
    "Age": [28, 22, 25, 30],
    "Salary": [50000, 70000, 45000, 60000]
}, index=["D", "B", "A", "C"])
df
````

### **axis**
- axis=0 → row index
- axis=1 → column index

````python
df.sort_index(axis=0)
````

## **NaN**

````python
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun"],
    "Age": [28, 22, 25]
}, index=["B", None, "A"])
````

````python
df.sort_index(na_position="first")
````

### **ignore_index**
ignore_index=False → keep index labels

ignore_index=True  → reset to 0, 1, 2...

````python
df.sort_index(ignore_index=True)
````

### **level**
MultiIndex

````python
df = pd.DataFrame({
    "Salary": [50000, 60000, 70000, 80000]
}, index=pd.MultiIndex.from_tuples([
    ("CSE", "Ravi"),
    ("ECE", "Priya"),
    ("CSE", "Arun"),
    ("ECE", "Sneha")
], names=["Branch", "Name"]))
````
