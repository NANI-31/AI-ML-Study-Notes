```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita", "Kiran", "Meena"],
    "Age": [24, 25, 23, 26, 24, 29],
    "Salary": [50000, 65000, 55000, 70000, 60000, 80000],
    "Branch": ["CSE", "ECE", "CSE", "ECE", "CSE", "IT"],
    "City": ["Vijayawada", "Hyderabad", "Chennai",
             "Hyderabad", "Vijayawada", "Bangalore"]
})
df
```

### **1. isin()**

```python
df[df["Branch"].isin(["CSE", "ECE"])]
df[df["Age"].isin([22,23,25,29])]
```

### **2. between()**

```python
df[df["Salary"].between(60000, 70000)]
df[df["Salary"].between(60000, 70000, inclusive="neither")]
# 60000 → excluded
# 70000 → excluded
```

### **3. query()**

```python
df[(df["Salary"] > 60000) & (df["Age"] > 25)]
df.query("Age > 25 and Salary > 60000")
```

### **Using Python variables with query()**

```python
min_sal = 60000
df.query("Salary >= @min_sal")
```

### **4. filter()**

DataFrame.filter() is primarily used to select labels, usually columns or indexes.

```python
df.filter(items=["Name", "Salary"])
```

```python
#  select columns containing "Score"
df.filter(like="Score")
# the column name ends with "score".
df.filter(regex="score$")
```

### **5. Boolean Masks**

```python
age_mask = df["Age"] >= 25
salary_mask = df["Salary"] > 60000
result = df[
    age_mask & salary_mask
]
result
```

### **6. Conditional Filtering with loc[]**
df.loc[condition, columns]

```python
df.loc[df["Salary"] > 60000, ["Name", "Salary"]]
```

```python
# Give everyone with salary below ₹60,000 a 10% raise
df.loc[df["Salary"] < 60000, "Salary"] = (
    df.loc[df["Salary"] < 60000, "Salary"] * 1.10
).astype(int)

# or

# df["Salary"] = df["Salary"].astype(float)
# df.loc[df["Salary"] < 60000, "Salary"] *= 1.10

# int64 column
#    ↓
# float64 values
#    ↓
# TypeError
df
```

```python
df.loc[
    (df["Branch"].isin(["CSE", "ECE"])) &
    (df["Age"].between(24, 26)) &
    (df["Salary"] >= 55000),
    ["Name", "Age", "Salary", "City"]
]
df.query(
    "Branch in ['CSE', 'ECE'] and "
    "Age >=24 and Age <= 26 and "
    "Salary >= 55000"
)[
    ["Name", "Age", "Salary", "City"]
]
```
