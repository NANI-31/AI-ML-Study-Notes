```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita", "Kiran", "Meena"],
    "Age": [24, 25, 23, 26, 24, 29],
    "Salary": [50000, 65000, 55000, 70000, 60000, 80000],
    "Branch": ["CSE", "ECE", np.nan, "ECE", "CSE", "IT"],
    "City": ["Vijayawada", "Hyderabad", "Chennai",
             "Hyderabad", "Vijayawada", "Bangalore"]
})
df
```

### **1. Filtering Rows**

```python
df[df["Age"] > 25]
```

```python
df[df["Salary"] > 60000]
```

### **2. Filtering Columns**

```python
df[["Name", "Salary"]]
```

```python
numeric_cols = df.select_dtypes(include="number")
numeric_cols
```

### **Filtering Rows AND Columns**

```python
df.loc[df["Salary"] > 60000, ["Name", "Salary"]]
```

### **3. Filtering Numerical Values**

```python
df[df["Age"] > 25]
df[df["Age"] >= 25]
df[df["Age"] < 25]
df[df["Age"] <= 25]
df[df["Age"] == 25]
df[df["Age"] == 25] / df[~(df["Age"] == 25)]
```

### **4. Filtering String Values**

```python
df[df["Branch"] == "CSE"]
df[~(df["Branch"] == "CSE")]
```

### **String methods**

```python
df[df["City"].str.contains("Hyd")]
df[df["Branch"].str.lower() == "cse"]
df[df["Branch"].str.contains("cse", case=False, na=True)]  # na, If the value is missing(Nulls), return False, without it, returns None/null/NaN
```

```python
df[df["Name"].str.startswith("R")]
df[df["Name"].str.endswith("a")]
```
