# Special Arrays
creating arrays with predefined values
- np.zeros()
- np.ones()
- np.empty()
- np.full()
- np.eye()
- np.identity()


```python
arr1 = np.zeros(5)
arr2 = np.zeros((2,2))
print(arr1)
print(arr2)
arr = np.zeros(5, dtype=int)
print(arr)
```

```python
arr1 = np.ones(5, dtype=int)
arr2 = np.ones((2,4), dtype = int)
print(arr1)
print(arr2)
```

```python
arr = np.empty(3)
print(arr)
```

```python
arr1 = np.full(5, 7) #(shape, value)
arr2 = np.full((2,3), 9)
print(arr1)
print(arr2)
```

```python
arr1 = np.eye(3) #identity like-matrix
arr2 = np.eye(3,5)
print(arr1)
print(arr2)
arr3 = np.eye(3, k=1) #offset
print(arr3)
```

```python
arr1 = np.identity(5)
print(arr1)
arr2 = np.identity(2,3)
print(arr2)
```
