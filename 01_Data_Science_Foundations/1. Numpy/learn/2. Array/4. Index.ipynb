`````python
import numpy as np
`````

`````python
a = np.arange(9).reshape(3,3)
print(a)
`````

`````python
print(a[-1,0])
`````

`````python
# selecting entire row
print(a[1])
#selecting enire column
print(a[:,1])
# means:
# : → all rows
# 0 → column 0
`````

`````python
#multi rows
print(a[0:2])
# multi cols
print(a[:, 1:3])
`````

`````python
# specific portion
print(a[0:2, 0:2])
`````

`````python
#  Get first two rows and last two columns
print(a[:2,:2])
#  Get last col
print(a[:, -1])
`````

`````python
# reverse the rows
print(a[::-1, :])
# reverse the cols
print(a[:, ::-1])
`````

`````python
# Take every other row
print(a[::2, :])
#  Take every other col
print(a[:,::2])
`````

| Indexing      | Meaning                       |
| ------------- | ----------------------------- |
| `a[0, 0]`     | First element                 |
| `a[0, :]`     | First row                     |
| `a[:, 0]`     | First column                  |
| `a[1:3, :]`   | Rows 1–2                      |
| `a[:, 1:3]`   | Columns 1–2                   |
| `a[1:3, 1:3]` | Rows 1–2 + columns 1–2        |
| `a[:2, 2:]`   | First 2 rows + last 2 columns |
| `a[-1, :]`    | Last row                      |
| `a[:, -1]`    | Last column                   |
| `a[::-1, :]`  | Reverse rows                  |
| `a[:, ::-1]`  | Reverse columns               |
| `a[::2, :]`   | Every 2nd row                 |
| `a[:, ::2]`   | Every 2nd column              |


````a[ rows , columns ]

So a[:, :] means the entire 2-D array.
