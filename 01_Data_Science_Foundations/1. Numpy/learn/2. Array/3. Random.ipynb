````python
import numpy as np
````

- np.random.rand()
- np.random.randn()
- np.random.randint()
- np.random.seed()


````python
# Generates random numbers from a uniform distribution between 0 and 1
print(np.random.rand(5))
print(np.random.rand(2,2))
````

````python
# Generates random numbers from a standard normal distribution.
# Standard normal distribution means approximately:
# mean = 0
# standard deviation = 1

print(np.random.randn(5))

# Unlike rand(), values can be:
# negative
# positive
# near zero
# occasionally much larger/smaller
````

````python
print(np.random.randint(1,10,5))
print(np.random.randint(1,80,(2,4)))
````

````python
np.random.seed(45)
print(np.random.rand(5))
````

````python
arr = np.arange(1,10)
print(arr)
````

````python
print(np.random.choice(arr))
print(np.random.choice(arr,size = 4))
print(np.random.choice(arr,size = 4, replace=True))
````

````python
np.random.shuffle(arr)
print(arr)
````

````python
shuffled = np.random.permutation(arr)
print(shuffled)
print(arr)
````

```
np.random.shuffle()
    ↓
modifies original array
np.random.permutation()
    ↓
returns a shuffled copy
```
