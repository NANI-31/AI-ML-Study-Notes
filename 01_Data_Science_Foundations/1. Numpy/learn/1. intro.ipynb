```python
import numpy as np
```

```python
#check numpy version
print(np.__version__)
```

```python
arr = np.array([10,20,30,40])
print(arr)
print(type(arr))
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6,7])
c = np.array([7,8,9])
print(a+c)
print(a+b)
```

### Array type

```python
arr = np.array([1,2,3], dtype=np.int32)
print(arr)
print(arr.itemsize)
```

```python
# multi D
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr)
```

```python
# numerical calculations
marks = np.array([10,20,30,40, 44])
print(marks)
print("mean", np.mean(marks))
print("median", np.median(marks))
print("std", np.std(marks))
print("min", np.min(marks))
print("max", np.max(marks))
```

```python
print(arr)
print(np.sqrt(arr))
print(np.sin(arr))
print(np.cos(arr))
print(np.log(arr))
print(np.exp(arr))
```

# Basics


```python
#dimensions
print(a, a.ndim)
print(arr, arr.ndim)
```

```python
#shape
print(a.shape)
print(arr.shape)
```

```python
#size
print(a.size)
print(arr.size)
```

- **shape** tells how many elements exist along each dimension, [ rows ] / [row, coulmns] / [ samples, features] 
- **size** tells the total number of elements
