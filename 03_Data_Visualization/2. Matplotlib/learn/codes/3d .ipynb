```python
x = [1, 2, 3, 4, 5]
y = [1, 3, 2, 5, 4]
z = [2, 4, 3, 6, 5]
fig = plt.figure()
ax = fig.add_subplot(
    111,
    projection="3d"
)
ax.plot(
    x,
    y,
    z
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title(
    "3D Line Plot"
)
plt.show()
```

```python
t = np.linspace(
    0,
    10,
    500
)
x = np.cos(t)
y = np.sin(t)
z = t
fig = plt.figure()
ax = fig.add_subplot(
    111,
    projection="3d"
)
ax.plot(
    x,
    y,
    z
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Time / Z")
ax.set_title(
    "3D Spiral Trajectory"
)
plt.show()
```

```python
t = np.linspace(
    0,
    10,
    500
)

x = np.cos(t)
y = np.sin(t)
z = t

fig, ax = plt.subplots()

ax.plot(
    x,
    y
)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("2D Spiral Trajectory")

plt.show()
```

```python
x = np.random.rand(200)
y = np.random.rand(200)
z = np.random.rand(200)
values = np.sqrt(
    x**2 + y**2 + z**2
)
fig = plt.figure()
ax = fig.add_subplot(
    111,
    projection="3d"
)
scatter = ax.scatter(
    x,
    y,
    z,
    c=values,
    cmap="viridis"
)
fig.colorbar(
    scatter,
    ax=ax,
    label="Magnitude"
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.show()
```

```python
x = np.array([0, 1, 2, 3])
y = np.zeros(4)
z = np.zeros(4)
dx = np.ones(4) * 0.5
dy = np.ones(4) * 0.5
dz = np.array([
    5,
    8,
    3,
    7
])
fig = plt.figure()
ax = fig.add_subplot(
    111,
    projection="3d"
)
ax.bar3d(
    x,
    y,
    z,
    dx,
    dy,
    dz
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Value")
plt.show()
```

```python
x = np.linspace(
    -5,
    5,
    30
)
y = np.linspace(
    -5,
    5,
    30
)
X, Y = np.meshgrid(
    x,
    y
)
Z = X**2 + Y**2
fig = plt.figure()
ax = fig.add_subplot(
    111,
    projection="3d"
)
ax.plot_wireframe(
    X,
    Y,
    Z
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
plt.show()
```

```python
x = np.linspace(
    -5,
    5,
    100
)
y = np.linspace(
    -5,
    5,
    100
)
X, Y = np.meshgrid(
    x,
    y
)
Z = X**2 + Y**2
fig = plt.figure(
    figsize=(10, 7)
)
ax = fig.add_subplot(
    111,
    projection="3d"
)
ax.plot_surface(
    X,
    Y,
    Z
)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Z")
ax.set_title(
    "3D Surface"
)
plt.show()
```
