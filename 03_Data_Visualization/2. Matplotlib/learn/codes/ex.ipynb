```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'discount': [
        0, 5, 10, 15, 20,
        25, 30, 5, 10, 20,
        15, 25, 0, 10, 20,
        30, 5, 15, 25, 10
    ],

    'profit': [
        200, 150, 300, 250, 180,
        220, 280, 190, 310, 170,
        260, 230, 210, 290, 160,
        270, 200, 240, 180, 300
    ],

    'sales': [
        100, 120, 150, 180, 200,
        220, 250, 280, 300, 350,
        400, 450, 500, 600, 750,
        900, 1200, 2000, 5000, 10000
    ],

    'category': [
        'Furniture', 'Furniture', 'Furniture', 'Furniture',
        'Furniture', 'Furniture', 'Furniture', 'Furniture',
        'Furniture', 'Furniture',

        'Technology', 'Technology', 'Technology', 'Technology',
        'Technology', 'Technology', 'Technology', 'Technology',
        'Technology', 'Technology'
    ]
}

df = pd.DataFrame(data)
df
```

```python
# Calculate correlation
correlation = df['discount'].corr(df['profit'])

print("Correlation:", correlation)

# Scatter plot
plt.scatter(df['discount'], df['profit'])
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs Profit')
plt.show()
```

```python
plt.hist(df['sales'], bins=30, edgecolor='black')

plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Distribution of Sales')

plt.show()
```

```python
import matplotlib.pyplot as plt

df.boxplot(column='profit', by='category')

plt.xlabel('Category')
plt.ylabel('Profit')
plt.title('Profit by Category')

plt.suptitle('')  # removes pandas' automatic title
plt.show()
```

```python
# Finding A: Correlation
correlation = df['discount'].corr(df['profit'])
print("Discount-Profit correlation:", correlation)

plt.scatter(df['discount'], df['profit'])
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs Profit')
plt.show()


# Finding B: Right-skewed sales
print("Sales skewness:", df['sales'].skew())

plt.hist(df['sales'], bins=30, edgecolor='black')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Distribution of Sales')
plt.show()


# Finding C: Profit by category
sns.boxplot(data=df, x='category', y='profit')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.title('Profit by Category')
plt.show()
```
