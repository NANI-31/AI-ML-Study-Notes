```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
hours = [1, 2, 3, 4, 5]
scores = [50, 55, 65, 70, 80]
plt.scatter(
    hours,
    scores
)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")
plt.show()
```

```python
age = [20, 25, 30, 35, 40]
salary = [25000, 35000, 45000, 60000, 75000]
plt.scatter(
    age,
    salary,
    label="no",
    s=100,
    marker="*",
    color="r"
)
plt.xlabel("Age")
plt.ylabel("Salary")
plt.legend()
plt.show()
```

```python
x = np.random.randn(1000)
y = np.random.randn(1000)
experience = np.random.randn(1000)
scatter = plt.scatter(
    x,
    y,
    # alpha=0.3,
    s=[x * 20 for x in experience],
    c=y,
    cmap="viridis"
)
plt.colorbar(scatter, label="FEature value")
plt.show()
```

```python
hours = [1, 2, 3, 4, 5]
scores = [
    50,
    60,
    65,
    75,
    85
]
age = [
    18,
    19,
    20,
    21,
    22
]
plt.scatter(
    hours,
    scores,
    c=age,
    cmap="viridis"
)
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title("Study Hours vs Score")
plt.colorbar(
    label="Age"
)
plt.show()
```

```python
age = [22, 25, 28, 32, 36]
salary = [
    30000,
    40000,
    50000,
    65000,
    80000
]
performance = [
    60,
    70,
    80,
    90,
    95
]
experience = [
    1,
    2,
    4,
    7,
    10
]
plt.scatter(
    age,
    salary,
    c=performance,
    s=np.array(experience) * 30,
    cmap="viridis",
    alpha=0.7
)
plt.colorbar(
    label="Performance"
)
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title(
    "Age vs Salary"
)
plt.show()
```

# Multiple Categories

```python
class_a_x = [1, 2, 3]
class_a_y = [2, 3, 4]
class_b_x = [5, 6, 7]
class_b_y = [6, 7, 8]
class_c_x = [2, 3, 4]
class_c_y = [8, 7, 9]
plt.scatter(
    class_a_x,
    class_a_y,
    label="Class A"
)
plt.scatter(
    class_b_x,
    class_b_y,
    label="Class B"
)
plt.scatter(
    class_c_x,
    class_c_y,
    label="Class C"
)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Class Distribution")
plt.legend()
plt.show()
```

Color Based on Values

```python
x = np.random.rand(100)
y = np.random.rand(100)
temperature = np.random.rand(100) * 40
scatter = plt.scatter(
    x,
    y,
    c=temperature,
    cmap="coolwarm"
)
plt.colorbar(
    scatter,
    label="Temperature"
)
plt.xlabel("X")
plt.ylabel("Y")
plt.title(
    "Points Colored by Temperature"
)
plt.show()
```

Size Based on Values

```python
x = np.random.rand(100)
y = np.random.rand(100)
population = np.random.randint(
    10,
    1000,
    100
)
plt.scatter(
    x,
    y,
    s=population / 10,
    alpha=0.5
)
plt.xlabel("X")
plt.ylabel("Y")
plt.title(
    "Point Size Represents Population"
)
plt.show()
```

Basic Trend Line

```python
x = np.array([
    1, 2, 3, 4, 5
])
y = np.array([
    50, 55, 65, 72, 80
])
plt.scatter(
    x,
    y
)
m, b = np.polyfit(
    x,
    y,
    1
)
trend = m * x + b
x_line = np.linspace(
    x.min(),
    x.max(),
    100
)
y_line = m * x_line + b
plt.plot(
    # x,
    # trend,
    x_line,
    y_line,

    linestyle="--"
)
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title(
    "Study Hours vs Score"
)
plt.show()
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([50, 55, 65, 72, 80])
plt.scatter(
    x,
    y,
    alpha=0.7
)
m, b = np.polyfit(
    x,
    y,
    1
)
x_line = np.linspace(
    x.min(),
    x.max(),
    100
)
y_line = m * x_line + b
plt.plot(
    x_line,
    y_line,
    linestyle="--",
    label="Trend"
)
plt.xlabel("Study Hours")
plt.ylabel("Score")
plt.title(
    "Study Hours vs Score"
)
plt.legend()
plt.show()
```

Highlighting Points

```python
x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([10, 15, 12, 20, 18, 30])
plt.scatter(
    x,
    y,
    alpha=0.5
)
# Highlight maximum value
index = np.argmax(y)
plt.scatter(
    x[index],
    y[index],
    alpha=0.2
    # s=200,
    # marker="*"
)
plt.xlabel("X")
plt.ylabel("Y")
plt.title(
    "Highlighted Maximum Point"
)
plt.show()
```

Highlighting an Outlier

```python
x = [1, 2, 3, 4, 5, 6]
y = [10, 12, 11, 13, 12, 50]
plt.scatter(
    x,
    y,
    alpha=0.5
)
plt.scatter(
    6,
    50,
    s=200,
    marker="*"
)
plt.show()
```

Annotating Points

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.scatter(
    x,
    y
)
plt.annotate(
    "Important Point",
    xy=(4, 30),
    xytext=(3, 35),
    arrowprops=dict(
        arrowstyle="->"
    )
)
plt.show()
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [10, 12, 11, 13, 12, 50]
plt.scatter(
    x,
    y
)
plt.annotate(
    "Possible Outlier",
    xy=(6, 50),
    xytext=(4.5, 55),
    arrowprops=dict(
        arrowstyle="->"
    )
)
plt.xlabel("X")
plt.ylabel("Y")
plt.title(
    "Outlier Detection"
)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
labels = [
    "A",
    "B",
    "C",
    "D",
    "E"
]
plt.scatter(
    x,
    y
)
for x_value, y_value, label in zip(x,y,labels):
    plt.annotate(
        label,
        (x_value, y_value),
        xytext=(5, 5),
        textcoords="offset points"
    )
plt.show()
```

```python
x = np.random.rand(100) * 10
y = (
    3 * x
    + np.random.randn(100) * 5
)
# Third variable
values = np.random.rand(100) * 100
# Fourth variable
sizes = np.random.rand(100) * 200 + 20
fig, ax = plt.subplots(
    figsize=(10, 6)
)
# Scatter
scatter = ax.scatter(
    x,
    y,
    c=values,
    s=sizes,
    cmap="viridis",
    alpha=0.6,
    edgecolors="black",
    linewidths=0.5
)
# Colorbar
fig.colorbar(
    scatter,
    # ax=ax,
    label="Value"
)
# Trend line
m, b = np.polyfit(
    x,
    y,
    1
)
x_line = np.linspace(
    x.min(),
    x.max(),
    100
)
y_line = (
    m * x_line + b
)
ax.plot(
    x_line,
    y_line,
    linestyle="--",
    label="Trend"
)
# Find maximum point
index = np.argmax(y)
ax.scatter(
    x[index],
    y[index],
    s=300,
    marker="*"
)
# Annotate maximum
ax.annotate(
    "Maximum",
    xy=(x[index], y[index]),
    xytext=(10, 10),
    textcoords="offset points",
    arrowprops=dict(
        arrowstyle="->"
    )
)
ax.set_xlabel(
    "Feature X"
)
ax.set_ylabel(
    "Feature Y"
)
ax.set_title(
    "Advanced Scatter Plot"
)
ax.legend()
plt.show()
```
