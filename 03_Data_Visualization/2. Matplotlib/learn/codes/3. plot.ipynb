```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.figure(dpi=100)
x = [1, 2, 3, 4, 5]
sales = [100, 150, 130, 180, 200]
expenses = [80, 100, 110, 120, 140]
plt.plot(x, sales)
plt.plot(x, expenses)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
plt.plot(x, y)
plt.title("Sine Function")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.cos(x)
plt.plot(x, y)
plt.title("Cosine Function")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 10)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1, label="sin",     marker="o",
    markeredgecolor="black",
    markeredgewidth=1,
    alpha=0.1
)
plt.plot(x, y2, label="cos")
plt.legend()
plt.show()
```

```python
x = np.linspace(-3, 3, 100)
y1 = x ** 2
y2 = x ** 3
plt.plot(
    x,
    y1,
    "ro-",
    label="x²",
)
line, = plt.plot(
    x,
    y2,
    label="x³",
    marker="$E$"
)
line.set_dashes([5,2,1,5])
plt.title("Mathematical Functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
