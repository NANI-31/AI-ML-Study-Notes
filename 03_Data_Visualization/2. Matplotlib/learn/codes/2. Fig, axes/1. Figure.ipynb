```python
import matplotlib.pyplot as plt
```

```python

# Graph 1 — WITHOUT tight_layout
fig1 = plt.figure(figsize=(5, 3))

plt.plot([1, 2, 3], [10, 20, 30])
plt.xlabel("This is a very long X-axis label")
plt.ylabel("This is a very long Y-axis label")
plt.title("Without tight_layout")

# plt.show()


# Graph 2 — WITH tight_layout
fig2 = plt.figure(figsize=(5, 3), tight_layout=True)

plt.plot([1, 2, 3], [10, 20, 30])
plt.xlabel("This is a very long X-axis label")
plt.ylabel("This is a very long Y-axis label")
plt.title("With tight_layout")
fig2.get_dpi()
# plt.show()
```

```python
fig, ax = plt.subplots()
fig.set_facecolor("lightgray")
ax.set_facecolor("red")
ax.plot([1,2,3], [10,20,30])
ax.set_title("training loss")
fig.suptitle("Machine learing")
plt.show()
```

```python
fig = plt.figure(figsize=(10, 6))
ax1 = fig.add_axes([0.1, 0.1, 0.4, 0.8])
ax2 = fig.add_axes([0.55, 0.1, 0.35, 0.8])
ax1.plot([1, 2, 3], [10, 20, 30])
ax1.set_title("Plot 1")
ax2.plot([1, 2, 3], [30, 20, 10])
ax2.set_title("Plot 2")
plt.show()
```
