```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
values = [40, 30, 20, 10]
plt.pie(values)
plt.show()
```

```python
categories = [
    "Training",
    "Validation",
    "Testing"
]
values = [
    70,
    15,
    15
]
plt.pie(
    values,
    labels=categories
)
plt.title("Dataset Split")
plt.show()
```

```python
labels = [
    "Model A",
    "Model B",
    "Model C"
]
values = [
    500,
    300,
    200
]
explode = [0.02, 0.02, 0.04]
plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    counterclock=False,
    startangle=90,
    explode=explode,
    shadow = True,
    radius=0.5
)
plt.axis("equal")
plt.title("Model Usage")
plt.show()
```

 Donut Charts

```python
labels = [
    "Training",
    "Validation",
    "Testing"
]
values = [
    70,
    15,
    15
]
plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    wedgeprops={
        "width": 0.4
    },
    startangle=90,
    counterclock=False,
    shadow=True
)
plt.axis("equal")
plt.show()
```

```python
labels = [
    "Training",
    "Validation",
    "Testing"
]
values = [
    70,
    15,
    15
]
fig, ax = plt.subplots()
ax.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    wedgeprops={
        "width": 0.4
    }
)
ax.text(
    0,
    0,
    "100%",
    ha="center",
    va="center",
    fontsize=18
)
ax.axis("equal")
plt.show()
```

Nested Pie

```python
outer_values = [
    60,
    25,
    15
]
inner_values = [
    40,
    20,
    15,
    10,
    10,
    5
]
fig, ax = plt.subplots()
ax.pie(
    outer_values,
    radius=1,
    wedgeprops={
        "width": 0.3,
        "edgecolor": "white"
    }
)
ax.pie(
    inner_values,
    radius=0.7,
    wedgeprops={
        "width": 0.3,
        "edgecolor": "white"
    }
)
ax.axis("equal")
plt.show()
```

```python
labels = [
    "Class A",
    "Class B",
    "Class C"
]
data_2024 = [
    50,
    30,
    20
]
data_2025 = [
    45,
    35,
    20
]
data_2026 = [
    40,
    40,
    20
]
fig, axes = plt.subplots(
    1,
    3,
    figsize=(12, 4)
)
axes[0].pie(
    data_2024,
    labels=labels,
    autopct="%1.0f%%"
)
axes[0].set_title("2024")
axes[1].pie(
    data_2025,
    labels=labels,
    autopct="%1.0f%%"
)
axes[1].set_title("2025")
axes[2].pie(
    data_2026,
    labels=labels,
    autopct="%1.0f%%"
)
axes[2].set_title("2026")
# plt.tight_layout()
plt.show()
```

Showing Both Percentage and Actual Value

```python
values = [
    700,
    150,
    150
]
labels = [
    "Training",
    "Validation",
    "Testing"
]
total = sum(values)
def format_label(percentage):
    actual = round(percentage * total / 100)
    return (
            f"{percentage:.1f}%\n"
            f"({actual})"
        )
plt.pie(
    values,
    labels=labels,
    autopct=format_label,
    pctdistance=0.8,
    rotatelabels=True
)
plt.axis("equal")
plt.show()
```
