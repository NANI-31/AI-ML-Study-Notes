```python
import matplotlib.pyplot as plt
import numpy as np
```

Instead of plotting every individual value, the histogram groups values into ranges called bins

```python
ages = [
    18, 20, 21, 22, 22,
    23, 24, 25, 25, 26,
    27, 28, 30, 32, 35
]
plt.hist(ages)
plt.show()
```

Number of Bins

Number of Bins Changes the Shape

```python
plt.hist(
    ages,
    bins=20
)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

Bin Width

Instead of specifying the number of bins, you can explicitly specify the bin boundaries.

```python
scores = [
    5, 8, 12, 15, 17,
    22, 25, 28, 31, 35,
    38, 42, 45, 48
]
bins = [0, 10, 20, 30, 40, 50]
plt.hist(
    scores,
    bins=bins
)
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.title("Score Distribution")
plt.show()
```

Getting Histogram Values

Y-axis represents probability density, not the raw number of observations

The total area under the histogram is approximately:1

This is useful when comparing datasets with different numbers of observations

```python
counts, bins, patches = plt.hist(
    ages,
    bins=5,
    density=True
)
print(counts, bins, patches)
```

```python
data = np.random.normal(
    50,
    10,
    1000
)
plt.hist(
    data,
    bins=30,
    density=True,
    alpha=0.7
)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Probability Distribution")
plt.show()
```

# Cumulative Histogram

```python
plt.hist(
    ages,
    bins=10,
    cumulative=True,
    # density=True
)
plt.show()
```

# Multiple Histograms

```python
group_a = [20, 21, 22, 23, 24, 25]
group_b = [25, 26, 27, 28, 29, 30]
bi = np.arange(0, 101, 10)
plt.hist(
    group_a,
    # bins=bi,
    # alpha=0.2
)
plt.hist(
    group_b,
    # bins=bi
)
plt.show()
```

# Stacked Histograms

```python
group_a = [20, 21, 22, 23, 24, 25]
group_b = [24, 25, 26, 27, 28, 29]
plt.hist(
    [group_a, group_b],
    bins=6,
    stacked=True,
    label=["Group A", "Group B"]
)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

# Normalized Histograms

```python
# dataset_a = np.random.normal(loc=50, scale=10, size=10000)
dataset_a = np.arange(10000)
dataset_b = np.arange(1000)
plt.hist(
    dataset_a,
    bins=30,
    # density=True,
    # alpha=0.5,
    label="Dataset A"
)
plt.hist(
    dataset_b,
    bins=30,
    # density=True,
    # alpha=0.5,
    label="Dataset B"
)
plt.legend()
plt.show()
```

```python
group_a = np.random.normal(
    50,
    10,
    1000
)
group_b = np.random.normal(
    60,
    10,
    1000
)
bins = np.linspace(
    10,
    100,
    30
)
plt.hist(
    group_a,
    bins=bins,
    density=True,
    alpha=0.5,
    label="Group A"
)
plt.hist(
    group_b,
    bins=bins,
    density=True,
    alpha=0.5,
    label="Group B"
)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Distribution Comparison")
plt.legend()
plt.show()
```
