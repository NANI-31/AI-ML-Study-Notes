```python
import matplotlib.pyplot as plt
import numpy as np
```

### Generating a Normal Distribution

```python
data = np.random.normal(
    loc=50,
    scale=10,
    size=10000
)
plt.hist(
    data,
    bins=50,
    density=True
)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

### Identifying Skewness

```python
data = np.random.exponential(
    scale=10,
    size=5000
)
plt.hist(
    data,
    bins=50
)
plt.title("Right-Skewed Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### Identifying Outliers

```python
data = np.concatenate([
    np.random.normal(50, 10, 1000),
    [150, 170, 200]
])
plt.hist(
    data,
    bins=50
)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution with Possible Outliers")
plt.show()
```

```python
data = np.random.normal(
    50,
    10,
    1000
)
mean = np.mean(data)
plt.hist(
    data,
    bins=30,
    edgecolor="black"
)
plt.axvline(
    mean,
    linestyle="--",
    label=f"Mean = {mean:.2f}"
)
plt.legend()
plt.show()
```

```python
data = np.random.normal(
    50,
    10,
    5000
)
plt.hist(
    data,
    bins=40,
    density=True,
    alpha=0.6
)
mean = np.mean(data)
std = np.std(data)
x = np.linspace(
    mean - 4 * std,
    mean + 4 * std,
    200
)
pdf = (
    1 / (std * np.sqrt(2 * np.pi)) * np.exp(-0.5 * ((x - mean) / std) ** 2)
)
plt.plot(
    x,
    pdf,
    linestyle="--",
    linewidth=2
)
plt.title(
    "Histogram with Normal Distribution"
)
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
