```python
import re
```

```python
text = "I have 10 apples and 20 oranges"

matches = re.finditer(r"\d+", text)
print(matches)
for m in matches:
    print(m.group())
    print(m)
```

```python
text = "Python Java Perl PHP bphp Ruby"

for m in re.finditer(r"P\w*", text, re.I):
    print(m.group())
```

```python
text = "cat dog cat bird cat"

matches = list(re.finditer(r"\bcat\b", text))
print(reversed(matches))
print(matches)
for m in reversed(matches):
    start, end = m.span()
    text = text[:start] + "CAT" + text[end:]

print(text)
```

```python
text = "10 50 101 200 99 500 20"
c = 0
def replace(match):
    global c
    n = int(match.group())

    if n > 100:
        c = c + 1
        return "BIG"

    return match.group()

result, count = re.subn(r"\d+", replace, text)

print(result)
print("Matches processed:", c)
```
