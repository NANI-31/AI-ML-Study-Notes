```python
import re
```

```python
def p(r):
    print(r)
    if r is None:
        return

    if isinstance(r, (list, str)):
        # print(f"findall items ({len(r)}) -> {r}")
        return

    if r.groupdict():
        print("groups ->", r.groups())
        print("named groups ->", r.groupdict())
        for name, val in r.groupdict().items():
            print(f"group:{name} -> {val}")
        return

    # for i in range(1, len(r.groups()) + 1):
    #     print(f"group:{i} -> {r.group(i)}")
    if r.groups():
        print("groups ->", r.groups())
        for i, val in enumerate(r.groups(), start=1):
            print(f"group:{i} -> {val}")

```

```python
text = "10% 20 30% 40"
r1 = re.findall(r"\b\d+\b(?!%)", text)
r = re.findall(r"\d+(?!%)", text)
p(r1)
p(r)
```

```python
text = "google.com example.org test.net amazon.com"

result = re.findall(
    r"\b\w+\.(?!com\b)\w+\b",
    text
)
result1 = re.findall(
    r"\b\w+\.(?!com\b).+\b",
    text
)

p(result)
p(result1)
```

```python

pattern = r"^(?!.*password).+$"

values = [
    "hello123",
    "mypassword",
    "Python123"
]

for value in values:
    print(value, bool(re.fullmatch(pattern, value)))
```

```python
text = "a1b2c3"

result = re.findall(
    r"(?!\d).",
    text
)
r = re.findall(r"\D", text)

print(result)
print(r)
```

```python
text = "cat dog catalog doghouse cat123 dog143 we"
r = re.search(r"\b(?!(cat|dog)\b)\w+\b",text)
p(r)
```

```python
t = " 123-4567"
pat = re.compile(r"(\d{3})-\d{4}", re.DEBUG)
r = pat.search(t)
p(r)
```
