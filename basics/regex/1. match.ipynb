```python
import re
```

```python
def p(r):
    if r:
        print(r)
        print(r.group())
    else: print(r)
```

```python
text = "hello python"

result1 = re.match("hello", text)
p(result1)

result2 = re.match("python", text)
p(result2)

result3 = re.search("python", text)
p(result3)
```

Character classes

```python
text = "123 a 1 bc123"
result = re.match(r"\d", text)
p(result)

result = re.match(r"\d+", text)
p(result)
```

Letters

```python
text = "python123"
r1 = re.match(r"[A-Za-z]+", text)
p(r1)

r2 = re.match(r"[A-Za-z]", text)
p(r2)

r3 = re.match(r"[0-9]", text)
p(r3)
```

\w generally matches:

letters
digits
underscore

```python
text = "python_123"
r1 = re.match(r"\w", text)
p(r1)

r2 = re.match(r"\w+", text)
p(r2)
```

```python
r = re.match(r"Hello$", "python Hello")
p(r)
```

Capturing groups

```python
# ask with and without "+"
text = "Nani 123"
r = re.match(r"(\w+) (\d+)", text)
p(r)
if r:
    print(r.group(1))
    print(r.group(2))
```

```python
print(re.match(r"\d+", ""))
```

```python
def validate(value):
    return re.match(r"\d{5}", value) is not None
    # return re.fullmatch(r"\d{5}", value)

print(validate("12345"))
print(validate("123abc"))
print(validate("12345abc"))
```

```python
x = re.match(r"\d+", "123abc")
y = re.match(r"\d*", "abc123")
print(x)
print(y)
```

```python
x = re.match(r"(\d+)-(\w+)", "123-abc")
if x:
    print(x.group())
    print(x.group(1))
    print(x.group(2))
```

# Interview coding problems

```python
# 11. Extract the username from a string: john123@gmail.com john123
email = "john123@gmail.com"
x = re.match(r"(\w+)", email)
print(x)

m = re.match(r"(\w+)@", email)
if m:
    print(m.group(1))
```

```python
# Extract name and age nani 23
text = "nani 23"
x = re.match(r"(\w+) (\d+)", text)  #give space
print(x)
if x:
    print("Name:", x.group(1))
    print("Age:", x.group(2))
```

```python
# Check whether a string starts with a number
text = "a123abc"
x = re.match(r"\d", text)
if x: print("true")
else: print("false")
```

```python
# Parse a log line
line = "ERROR 404 NotFound"

m = re.match(r"(\w+) (\d+) (.+)", line)

if m:
    level = m.group(1)
    code = m.group(2)
    message = m.group(3)

    print(level)
    print(code)
    print(message)
```
