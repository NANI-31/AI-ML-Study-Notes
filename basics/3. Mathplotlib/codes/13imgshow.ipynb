```python
image = np.zeros(
    (300, 300, 3),
    dtype=np.uint8
)
# Top-left → Red
image[:150, :150, 0] = 255
# Top-right → Green
image[:150, 150:, 1] = 255
# Bottom-left → Blue
image[150:, :150, 2] = 255
# Bottom-right → White
image[150:, 150:] = 255
plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
def render_values(data, ax):
    for i in range(data.shape[0]):
        for j in range(data.shape[1]):
            ax.text(
                j,
                i,
                data[i, j],
                ha="center",
                va="center",
                color="white"
            )
# Image 1: values from 0 to 100
image1 = np.array([
    [0, 20, 40],
    [60, 80, 100]
])

# Image 2: values from 40 to 60
image2 = np.array([
    [40, 45, 50],
    [50, 55, 60]
])

fig, axes = plt.subplots(1, 2, figsize=(8, 4))

# Image 1
im1 = axes[0].imshow(image1, cmap="viridis",  vmin=0, vmax=100)
axes[0].set_title("Image 1")

# Image 2
im2 = axes[1].imshow(image2, cmap="viridis", vmin=0, vmax=100)
axes[1].set_title("Image 2")

render_values(image1, axes[0])
render_values(image2, axes[1])

fig.colorbar(im1, ax=axes[0])
fig.colorbar(im2, ax=axes[1])

plt.show()
```

```python
x = np.linspace(
    -5,
    5,
    100
)
y = np.linspace(
    -5,
    5,
    100
)
X, Y = np.meshgrid(
    x,
    y
)
Z = X**2 + Y**2
plt.contourf(
    X,
    Y,
    Z,
    levels=20,
    cmap="viridis"
)
plt.colorbar(
    label="Z value"
)
plt.xlabel("X")
plt.ylabel("Y")
plt.title(
    "Filled Contour Plot"
)
plt.show()
```
