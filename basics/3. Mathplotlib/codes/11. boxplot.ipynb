```python
data = [
    10,
    12,
    15,
    18,
    20,
    22,
    25,
    30,
    41
]
box =plt.boxplot(
    data,
    patch_artist=True
)
for patch in box["boxes"]:
    patch.set_facecolor("skyblue")
    # patch.set_color("red")
    patch.set_linewidth(5)
for median in box["medians"]:
    median.set_color("red")
    median.set_linewidth(4)
for whisker in box["whiskers"]:
    whisker.set_color("black")
    whisker.set_linewidth(2)
    whisker.set_linestyle("--")
for flier in box["fliers"]:
    flier.set(
        marker="o",
        markersize=6,
        markerfacecolor="red",
        markeredgecolor="black"
    )

plt.show()
```

```python
age = [
    20, 22, 24, 25, 28,
    30, 32, 35, 40
]
salary = [
    25000, 30000, 32000,
    35000, 40000, 45000,
    50000, 60000, 70000
]
experience = [
    1, 2, 2, 3, 4,
    5, 6, 8, 10
]
data = [
    age,
    salary,
    experience
]
plt.boxplot(
    data,
    tick_labels=[
        "Age",
        "Salary",
        "Experience"
    ]
)
plt.title(
    "Feature Distributions"
)
plt.show()
```

```python

linear_errors = [
    -5, -3, -2, 0, 2, 4, 5
]
forest_errors = [
    -3, -2, -1, 0, 1, 2, 3
]
nn_errors = [
    -4, -2, -1, 1, 2, 3, 4
]
errors = [
    linear_errors,
    forest_errors,
    nn_errors
]
plt.boxplot(
    errors,
    tick_labels=[
        "Linear Regression",
        "Random Forest",
        "Neural Network"
    ]
)
plt.ylabel("Prediction Error")
plt.title(
    "Model Error Distributions"
)
plt.show()
```

```python
data = [
    [10, 12, 15, 18, 20, 22, 25, 30],
    [20, 22, 25, 28, 30, 35, 40, 45],
    [5, 8, 10, 12, 15, 18, 20, 35]
]
fig, ax = plt.subplots(
    figsize=(9, 6)
)
box = ax.boxplot(
    data,
    patch_artist=True,
    notch=True,
    tick_labels=[
        "Dataset A",
        "Dataset B",
        "Dataset C"
    ]
)
# Different box colors
colors = [
    "skyblue",
    "lightgreen",
    "lightcoral"
]
for patch, color in zip(
    box["boxes"],
    colors
):
    patch.set_facecolor(color)
# Median styling
for median in box["medians"]:
    median.set_color("red")
    median.set_linewidth(2)
# Whisker styling
for whisker in box["whiskers"]:
    whisker.set_linestyle("--")
# Outlier styling
for flier in box["fliers"]:
    flier.set(
        marker="o",
        markersize=5
    )
ax.set_ylabel("Value")
ax.set_title(
    "Comparison of Data Distributions"
)
ax.grid(
    axis="y",
    alpha=0.3
)
plt.tight_layout()
plt.show()
```
