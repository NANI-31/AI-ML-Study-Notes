```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.plot(
    x,
    y
)
plt.fill_between(
    x,
    y
)
plt.show()
```

```python
x = np.linspace(
    0,
    10,
    100
)
y = x ** 2
plt.plot(
    x,
    y
)
plt.fill_between(
    x,
    y
)
plt.show()
```

```python
x = np.linspace(
    0,
    10,
    100
)
y1 = x
y2 = x + 5
plt.plot(
    x,
    y1,
    label="Lower"
)
plt.plot(
    x,
    y2,
    label="Upper"
)
plt.fill_between(
    x,
    y1,
    y2,
    color="#7f7f7f",
    alpha=0.5
)
plt.legend()
plt.show()
```

```python
x = np.linspace(
    0,
    10,
    100
)
y1 = np.sin(x)
y2 = np.cos(x)
plt.fill_between(
    x,
    y1,
    alpha=0.4,
    label="Sine"
)
plt.fill_between(
    x,
    y2,
    alpha=0.4,
    label="Cosine"
)
plt.plot(
    x,
    y1
)
plt.plot(
    x,
    y2
)
plt.legend()
plt.show()
```

Positive and Negative Regions

Filling Only Above the X-Axis

Sometimes you don't want to fill negative values.

```python
x = np.linspace(
    0,
    10,
    200
)
y = np.sin(x)
plt.plot(
    x,
    y
)
plt.fill_between(
    x,
    y,
    where=(y >= 0),
    alpha=0.4
)
plt.fill_between(
    x,
    y,
    where=(y < 0),
    alpha=0.4
)
plt.axhline(
    0,
    linestyle="--"
)
plt.show()
```

```python
x = np.linspace(
    0,
    10,
    100
)
prediction = np.sin(x)
lower = prediction - 0.2
upper = prediction + 0.2
plt.plot(
    x,
    prediction,
    label="Prediction"
)
plt.fill_between(
    x,
    lower,
    upper,
    alpha=0.3,
    label="Uncertainty"
)
plt.xlabel("Input")
plt.ylabel("Prediction")
plt.title(
    "Prediction with Uncertainty"
)
plt.legend()
plt.show()
```

```python
months = [
    "Jan",
    "Feb",
    "Mar",
    "Apr",
    "May"
]
sales_a = [
    10,
    15,
    20,
    25,
    30
]
sales_b = [
    5,
    10,
    15,
    20,
    25
]
sales_c = [
    3,
    5,
    8,
    10,
    12
]
plt.stackplot(
    months,
    sales_a,
    sales_b,
    sales_c
)
plt.show()
```

```python
months = [
    "Jan",
    "Feb",
    "Mar",
    "Apr",
    "May",
    "Jun"
]
product_a = [
    20,
    25,
    30,
    35,
    40,
    45
]
product_b = [
    15,
    20,
    18,
    25,
    30,
    35
]
product_c = [
    10,
    12,
    15,
    18,
    20,
    25
]
plt.stackplot(
    months,
    product_a,
    product_b,
    product_c,
    labels=[
        "Product A",
        "Product B",
        "Product C"
    ]
)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title(
    "Product Sales Over Time"
)
plt.legend()
plt.show()
```

Normalized Stacked Area Chart


```python
product_a = np.array([
    20, 25, 30, 35
])
product_b = np.array([
    10, 15, 20, 15
])
product_c = np.array([
    10, 10, 10, 20
])
total = (
    product_a
    + product_b
    + product_c
)
a_pct = product_a / total
b_pct = product_b / total
c_pct = product_c / total
x = [
    "Jan",
    "Feb",
    "Mar",
    "Apr"
]
plt.stackplot(
    x,
    a_pct,
    b_pct,
    c_pct,
    labels=[
        "A",
        "B",
        "C"
    ]
)
plt.ylabel("Proportion")
plt.title(
    "Normalized Composition"
)
plt.legend()
plt.show()
```

# Use Cases

Time-Series Visualization

```python
days = [
    1, 2, 3, 4, 5,
    6, 7
]
users = [
    100,
    120,
    150,
    180,
    200,
    230,
    250
]

plt.plot(
    days,
    users
)
plt.fill_between(
    days,
    users,
    alpha=0.3
)
plt.xlabel("Day")
plt.ylabel("Users")
plt.title(
    "Users Over Time"
)
plt.show()
```

Time Series: Website Traffic

```python
days = np.arange(1, 31)
traffic = (
    1000
    + days * 50
    + np.random.randint(
        -100,
        100,
        30
    )
)
plt.plot(
    days,
    traffic
)
plt.fill_between(
    days,
    traffic,
    alpha=0.3
)
plt.xlabel("Day")
plt.ylabel("Visitors")
plt.title(
    "Website Traffic"
)
plt.show()
```

```python
epochs = [
    1,
    2,
    3,
    4,
    5
]
loss = [
    1.5,
    1.2,
    0.9,
    0.7,
    0.5
]
plt.plot(
    epochs,
    loss
)
plt.fill_between(
    epochs,
    loss,
    alpha=0.3
)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title(
    "Training Loss"
)
plt.show()
```

```python
epochs = [
    1, 2, 3, 4, 5, 6
]
train_loss = [
    1.5,
    1.2,
    0.9,
    0.7,
    0.5,
    0.4
]
val_loss = [
    1.6,
    1.3,
    1.0,
    0.9,
    1.0,
    1.2
]
plt.plot(
    epochs,
    train_loss,
    label="Training Loss"
)
plt.plot(
    epochs,
    val_loss,
    label="Validation Loss"
)
plt.legend()
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title(
    "Training vs Validation Loss"
)
```

```python
days = [
    1,
    2,
    3,
    4
]
daily = [
    10,
    20,
    15,
    25
]
cumulative = np.cumsum(
    daily
)

plt.plot(
    days,
    cumulative
)
plt.fill_between(
    days,
    cumulative,
    alpha=0.3
)
plt.xlabel("Day")
plt.ylabel("Cumulative Value")
plt.title(
    "Cumulative Data"
)
plt.show()
```

```python
time = np.arange(1, 11)
cpu = [
    20, 25, 30, 35, 40,
    45, 50, 55, 60, 65
]
memory = [
    10, 15, 20, 20, 25,
    30, 35, 35, 40, 45
]
network = [
    5, 10, 8, 12, 15,
    18, 20, 22, 25, 30
]
plt.stackplot(
    time,
    cpu,
    memory,
    network,
    labels=[
        "CPU",
        "Memory",
        "Network"
    ],
    alpha=0.8
)
plt.xlabel("Time")
plt.ylabel("Usage")
plt.title(
    "System Resource Usage"
)
plt.legend(
    loc="upper left"
)
plt.show()
```

```python
np.random.seed(42)
x = np.linspace(
    0,
    10,
    100
)
prediction = (
    2 * x
    + np.sin(x)
)
uncertainty = (
    0.5
    + 0.1 * x
)
lower = (
    prediction
    - uncertainty
)
upper = (
    prediction
    + uncertainty
)
plt.plot(
    x,
    prediction,
    label="Prediction"
)
plt.fill_between(
    x,
    lower,
    upper,
    alpha=0.3,
    label="Uncertainty"
)
plt.xlabel("Input")
plt.ylabel("Prediction")
plt.title(
    "Prediction with Uncertainty"
)
plt.legend()
plt.show()
```
