```python
x = [1,2,3,4,5,6]
y = [0,10,5,15, 10,20]
plt.plot(x,y)
plt.title("Test")
plt.xlabel("x-series")
plt.ylabel("y-series")
plt.xlim(0, 7)
plt.ylim(0, 21)
plt.text(6, 20, "Highest")
plt.plot(x,y,color='red', linewidth=2, linestyle="-", marker="o", markersize=8, alpha=0.8, label="Sales")
plt.legend(loc="upper right")
plt.grid(axis="y")
plt.show()
```

```python
x = [1,5,7,8]
sales = [2,3,4,5]
profit = [4,3,2,6]
plt.plot(x,sales, label = "sales", marker = "o")
plt.plot(x,profit, label = "profit", marker="o")
plt.legend()
# plt.xticks([2, 4, 6, 8,10,12])
# plt.xticks([1, 2, 3],["Jan", "Feb", "Mar"])
plt.xticks(rotation=45)
plt.annotate("Highest", xy=(5, 5), xytext=(5, 6), arrowprops={"arrowstyle": "->"})
plt.axhline(y=7)
plt.axvline(x=9)
plt.fill_between(x, sales, alpha=0.4)
plt.axis("tight")
plt.show()
```

```python
plt.subplot(2, 2, 1)
plt.plot(x,sales)
plt.subplot(2,2,2)
plt.plot(x,profit)
plt.subplot(2,2,3)
plt.hist(sales)
```

```python
fig, ax = plt.subplots(2,2)
ax[0,0].plot(x, sales)
ax[0,1].bar(x, sales)
ax[1,0].scatter(x, sales)
ax[1,1].hist(x)
plt.savefig("chart.jpg")
plt.figure(figsize=(80,50))
```

```python
import numpy as np
x = np.linspace(0,10,100)
y = np.sin(x)
plt.figure(figsize=(8, 4))
plt.plot(x,y)
plt.title("Sine Wave")
plt.savefig("results.png", dpi=300)
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")
x = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])
z = np.array([5, 10, 15, 20])
ax.scatter(x, y, z)
plt.show()
```

```python
import plotly.graph_objects as go

x = [1, 2, 3, 4, 5, 6]
y = [0, 10, 5, 15, 10, 20]
z = [2, 5, 8, 12, 15, 20]

fig = go.Figure()

fig.add_trace(go.Scatter3d(
    x=x,
    y=y,
    z=z,
    mode='lines+markers',
    line=dict(color='yellow', width=5),
    marker=dict(size=6)
))

fig.update_layout(
    title="Interactive 3D Graph",
    scene=dict(
        xaxis_title="X-axis",
        yaxis_title="Y-axis",
        zaxis_title="Z-axis"
    )
)

fig.show()
```
