```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Arun", "Sita", "Kiran"],
    "Age": [24, 25, 23, 26, 24],
    "Salary": [50000, 65000, 55000, 70000, 60000],
    "Branch": ["CSE", "ECE", "CSE", "ECE", "CSE"]
})
print(df)
```

### **1. Selecting a Single Column**

```python
df['Age']
```

```python
df['Age'].value_counts()
```

### **2. Selecting Multiple Columns**

Why two brackets?
- The outer brackets are DataFrame indexing:   `df[ ... ]`
- The inner brackets create a Python list:   `["Name", "Age"]`


```python
df[["Name", "Age"]]
```

### **Single vs Multiple Columns**
- `df["Age"]` → Series
- `df[["Age"]]` → DataFrame

```python
df[["Age"]]
```

```python
print(type(df["Age"]), df["Age"].shape)
print(type(df[["Age"]]), df[["Age"]].shape)
```

### **3. Selecting Columns Using Lists**

```python
selected_columns = ["Name", "Salary"]
result = df[selected_columns]
print(result, type(result), sep='\n')
```

### **4. Selecting Columns Dynamically**

```python
# Give me the columns whose data is numeric.
numerical_columns = df.select_dtypes(include="number")
numerical_columns
```

```python
# Give me the names of the columns that contain numeric data.
numerical_columns.columns
```

### **Columns containing a word**

Suppose you have:
|columns|
|--|
math_score
science_score
english_score
age
name


```python
score_columns = [col for col in df.columns if "score" in col]
df[score_columns]
```

### **or filter()**

```python
score_columns = df.filter(like="score")
```
