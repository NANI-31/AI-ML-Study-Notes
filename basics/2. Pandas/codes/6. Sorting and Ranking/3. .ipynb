```python
import pandas as pd
employees = pd.DataFrame({
    "Name": [
        "Ravi", "Priya", "Arun",
        "Sita", "Kiran", "Meena"
    ],
    "Department": [
        "Engineering", "HR", "Engineering",
        "HR", "Engineering", "Sales"
    ],
    "Experience": [
        3, 5, 2, 7, 4, 6
    ],
    "Salary": [
        60000, 70000, 55000,
        90000, 65000, 85000
    ],
    "Performance": [
        82, 91, 75, 95, 88, 90
    ]
})
employees
```

```python
# Highest salary first
employees.nlargest(1, "Salary")
employees.sort_values(by="Salary", ascending=False)
```

```python
# Highest performance first
employees.sort_values(by="Performance", ascending=False)
```

```python
# Top 3 performers
employees.nlargest(3, "Performance")
```

```python
# Lowest 3 salaries
employees.nsmallest(3, "Salary")
```

```python
# Rank employees by performance
employees["Performance_Rank"] = employees["Performance"].rank(ascending=False).astype(int)
employees
```

```python
# Rank within department
employees["Department_Rank"] = (
    employees
    .groupby("Department")["Performance"]
    .rank(ascending=False)
    .astype(int)
)
employees
```

```python
# Find top employee in each department
sorted_department = ( employees
    .sort_values(["Department", "Performance"], ascending=[True, False])
)
sorted_department
```

```python
top_per_department = ( sorted_department
    .groupby("Department")
    .head(1)
)
top_per_department
```

```python
# Sorting + Ranking Together
leaderboard = employees.copy()
leaderboard = ( leaderboard
    .sort_values(by="Performance", ascending=False)
)
leaderboard
# first rank it. then sort with that rank column
```
