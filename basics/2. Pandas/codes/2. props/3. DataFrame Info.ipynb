### **5. memory_usage()**
- memory_usage() tells you approximately how much memory each column uses.
- For columns containing strings/objects, `deep=True` tells Pandas to inspect the actual memory consumed by the objects rather than just counting the references.

````python
print(df.memory_usage())
print(df.memory_usage(deep=True))
df.memory_usage(deep=True).sum()
````

### **6. count()**
count() returns the number of non-missing values.

````python
print(df.count())
df.loc[2, "Marks"] = None
print(df.count())
````
