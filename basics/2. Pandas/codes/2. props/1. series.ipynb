```python
import pandas as pd
```

Series Properties

```python
marks = pd.Series(
    [85, 92, 76, 88, 95],
    index=["Ravi", "Priya", "Arun", "Sita", "Kiran"],
    name="Marks"
)
print(marks)
```

index
- The index property tells you the labels associated with each value.

```python
print(marks.index)
```

values
- The values property returns the underlying values.

```python
marks.values
```

```python
print(type(marks.values))
```

dtype

```python
print(marks.dtype)
```

```python
names = pd.Series(["Ravi", "Priya", "Arun"])
print(names.dtype) #str/ object based on Pandas version/settings
```

shape
- shape tells you the dimensions of the Series.

```python
print(marks.shape)
```

size
- size tells you the total number of elements.

```python
print(marks.size)
```

Difference between shape and size
- shape = structure/dimensions
- size  = total number of elements


name
- A Series can have a name

```python
print(marks.name)
```

```python
marks.name = "Exam Marks"
print(marks.name)
```
