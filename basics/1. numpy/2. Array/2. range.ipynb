```python
import numpy as np
```

## np.arange(start, stop, step)
### similar to range()

```python
arr = np.arange(0,10)
print(arr)
```

```python
arr = np.arange(10,0,-1)
print(arr)
```

## np.linspace(start, stop, num)
#### creates a specified number of evenly spaced values between two endpoints.

```python
arr = np.linspace(1,10,4)
print(arr)
```

## np.logspace()
#### generates values that are evenly spaced on a logarithmic scale.

```python
arr = np.logspace(0,3,4)
print(arr)
```

- 10^0 = 1
- 10^1 = 10
- 10^2 = 100
- 10^3 = 1000


```python
arr = np.logspace(0,5,6, dtype = int)
print(arr)
```

### arange  → fixed interval
### linspace → fixed number of evenly spaced points
### logspace → fixed number of logarithmically spaced points
