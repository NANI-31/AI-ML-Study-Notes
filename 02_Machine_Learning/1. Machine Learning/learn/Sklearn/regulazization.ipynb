```python
import pandas as pd
from sklearn.linear_model import Lasso, LinearRegression, Ridge
from sklearn.model_selection import train_test_split
```

```python
data = {
  "square_feet": [1200, 1500, 1800, 2100, 2400, 2700, 3000, 3300],
  "num_rooms": [3, 4, 4, 5, 5, 6, 7, 8],
  "age_years": [10, 8, 5, 4, 3, 2, 1, 1],
  "distance_center_km": [4, 6, 8, 10, 12, 14, 16, 18],
  "price_lakhs": [50, 65, 78, 95, 110, 125, 145, 160]
}

df = pd.DataFrame(data) 
X = df[['square_feet', 'num_rooms', 'age_years', 'distance_center_km']]
y = df['price_lakhs']

X_train, X_test, y_train, y_test = train_test_split(
  X, y, test_size=0.25, random_state=42
)


models = {
  'Linear Regression': LinearRegression(),
  'Ridge Regression': Ridge(alpha=1.0),
  'Lasso Regression': Lasso(alpha=0.5),
}


results = {}
for name, model in models.items():
  model.fit(X_train, y_train)
  r2 = model.score(X_test, y_test)
  coef_dict = dict(zip(X, model.coef_))
  results[name] = {'test_r2': r2, 'coefficients': coef_dict}


# formatted_output = '\n'.join(
#   f"{name}: \n R2 = {metrics['test_r2']},\n Coefficients ="
#   f" {metrics['coefficients']}\n"
#   for name, metrics in results.items()
# )
formatted_output = pd.DataFrame([
    {
        "Model": name,
        "R2": metrics["test_r2"],
        "Coefficients": metrics["coefficients"]
    }
    for name, metrics in results.items()
])

print(formatted_output.to_string(index=False))
```
